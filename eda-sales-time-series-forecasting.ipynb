```r
library(tidyverse)
```

```r
minmax_date <- function(df, date="date") {
    df %>%
        select(date) %>%
        summarise(`Date min` = min(date),
              `Date max` = max(date)
        )    
}
```

# Holidays Events

```r
df_holidays_events <- read_csv("/kaggle/input/competitions/store-sales-time-series-forecasting/holidays_events.csv")
df_holidays_events %>%
    head()
```

```r
df_holidays_events %>%
    minmax_date()
```
